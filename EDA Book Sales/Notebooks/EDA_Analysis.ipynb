# Exploratory Data Analysis

```python
# import requied libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the dataset into a pandas
df=pd.read_csv(r"F:\DS & DA Projects\real-time projects\EDA Book Sales\Data\Books_Data_Clean.csv")
df
```

```python
# display the first records in the df
df.head()
```

```python
## display the lst records in the df
df.tail()
```

```python
# display the statistical reports
df.describe().T
```

```python
# display the info
df.info()
```

```python
# keep books published after  1900
df=df[df['Publishing Year'] > 1900]
df
```

```python
# checking the total null or missing values in the df
df.isna().sum()
```

```python
df.dropna(subset ='Book Name', inplace = True)
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df.nunique()
```

```python
plt.hist(df['Publishing Year'])
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Year")
plt.show()
```

```python
df['genre'].value_counts().plot(kind='bar')
plt.xlabel('genre')
plt.ylabel('Number Of Books')
plt.title('Number Of Books In Each Genre')
plt.show()
```

```python
df.groupby("Author")["Book_average_rating"].mean().sort_values(ascending=False)
```

```python
sns.boxplot(x="genre", y="Book_average_rating", data=df)
plt.xlabel("Genre")
plt.ylabel("Book Rating Count")
plt.title("Box Plot of Book Rating Count For Each Genre")
plt.show()
```

```python
plt.scatter(df["sale price"], df["units sold"])
plt.xlabel(" Sale Price")
plt.ylabel("Units Sold")
plt.title(" Scatter Plot Of Sale vs Units Sold")
plt.show()
```

```python
df["language_code"].value_counts().sort_values(ascending=True)
```

```python
language_counts=df["language_code"].value_counts()
language_counts
```

```python
plt.pie(language_counts, labels=language_counts.index, startangle=90)
plt.title("language Distribution of Books")
```

```python
df.groupby("Publisher ")["publisher revenue"].sum().sort_values(ascending=False)
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].mean().sort_values(ascending=False).max()
```

```python
df.groupby("language_code").size().sort_values(ascending=False)
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].mean()
```

```python
plt.scatter(df["Book_average_rating"],df["Book_ratings_count"])
plt.xlabel("Book average rating")
plt.ylabel("Book ratings count")
plt.title("Scatter Plot of Book Average Rating vs Book Ratings Count")
plt.show()
```

```python
tatol_gross_sales_by_author=df.groupby("Author")["gross sales"].sum()
tatol_gross_sales_by_author.sort_values(ascending=False).head(20).plot(kind="bar")
plt.xlabel("Author")
plt.ylabel("Total gross Sales")
plt.title("Total Gross Sales For Each Author")
plt.show()
```

```python
sns.boxplot(x="Author_Rating", y="units sold", data=df)
plt.xlabel("Author Rating")
plt.ylabel("Units sold")
plt.title("Box Plot of Units Sold Fo Each Author Rating")
plt.show()
```
